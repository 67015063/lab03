Activity 1 : ให้นักศึกษาเขียนโปรแกรม

1. สร้าง Class Stack แล้ว
2. Push เก็บข้อมูล ABCDEF
3. แล้ว Pop ออกจน empty

```python
class Stack:
  def __init__(self,list=None):
    if list == None:
      self.items = []
    else :
      self.items = list

  def push(self, i):
    self.items.append(i)
    return(self.items)

  def pop(self):
    return self.items.pop()

s = Stack()
for i in 'ABCDEF' :
  s.push(i)
print(s.items)

for j in range(len(s.items)) :
  s.pop()
print(s.items)
```

Activity 2 : ให้นักศึกษาเขียนโปรแกรม

1. สร้าง Queue แล้ว
2. เก็บข้อมูล ABCDEF
3. หลังจากนั้นให้แสดงผล ขนาดของข้อมูลที่เก็บใน Queue

```python
class Queue:
  def __init__(self,list=None):
      if list == None:
        self.items = []
      else :
        self.items = list

  def enQueue(self,i):
    self.items.append(i)

  def deQueue(self):
    return self.items.pop(0)

  def size(self):
    return len(self.items)

q = Queue()
for i in 'ABCDEF' :
  q.enQueue(i)
print(q.items)

print(q.size())
```

Activity 3 : ให้นักศึกษาเขียนโปรแกรม

1. สร้าง Link List แล้ว
2. เก็บข้อมูล A,B,C,D,E,F แต่ละตัวอักษรอยู่ใน 1 node
3. หลังจากนั้นให้แทรกข้อมูล X,Y,Z ระหว่าง C,D

```python
class node:
  def __init__(self, data, next = None):
    self.data = data
    if next == None:
      self.next = None
    else:
      self.next = next

  def __str__(self):
    return str(self.data)

class creat_list:
  def __init__(self):
    dummy_node = node(data = None, next = None)
    self.head = dummy_node

  def append(self, data):
    p = node(data)
    if self.head == None:
      self.head = p
    else:
      t = self.head
      while t.next != None:
        t = t.next
      t.next = p

  def insert_after(self, data, next = None):
    if next == None:
      p = self.head
    else :
       p = self.head.next
    while p.data != None and p.data != next:
      p = p.next

    on = p.next
    nn = node(data)
    p.next = nn
    nn.next = on

  def delete_after(self, data):
    if next == None:
      p = self.head
    else :
      p = self.head.next
      while p is not None and p.data != data:
        p = p.next

    if p is None:
      print("Node not found")
      return

    if p.next is None:
      print("No node to delete after", data)
      return

    p.next = p.next.next

  def show(self):
    p = self.head.next
    while p :
      print(f"{p.data}",end=" : " if p.next else "\n")
      p = p.next

li = creat_list()
li.append("A")
li.append("B")
li.append("C")
li.append("D")
li.append("E")
li.append("F")
print("append = ",end=""),li.show()

li.insert_after('Z', 'C')
li.insert_after('Y', 'C')
li.insert_after('X', 'C')
print("insert after = ",end=""),li.show()

li.delete_after('C')
print("delete after = ",end=""),li.show()
```
